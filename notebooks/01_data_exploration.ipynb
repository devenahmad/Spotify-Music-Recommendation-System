## Load data

Source: TidyTuesday 2020-01-21 (Spotify songs).

```python
URL = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-01-21/spotify_songs.csv"

df = pd.read_csv(URL)
print("Shape:", df.shape)
df.head()
```

First look at the raw data: column types, non-null counts and summary statistics. The dataset has 32,833 rows and 23 columns.

```python
df.info()
```

```python
df.describe(include='all').reset_index(drop=True)
```

## Pre-Processing

Cleaning steps: missing values, duplicate songs, data types and text columns.

### Missing Values

Only `track_name`, `track_artist` and `track_album_name` have missing values (5 rows each). A song without a name cannot be shown as a recommendation, so these rows are dropped (32,833 -> 32,828).

```python
print("Missing values per column:")
print(df.isna().sum())

# Then drop all rows that contain NaN values (they are < 2% of data)
df = df.dropna().reset_index(drop=True)

print("\nShape after removing missing values:", df.shape)
```

### Duplicate Songs

There are no fully identical rows, but the same song appears in several playlists, so `track_id` repeats. Only the first row per `track_id` is kept, so every song appears once and cannot be recommended several times (32,828 -> 28,352).

```python
# Remove duplicate songs (same song appears in multiple playlists)
print("Shape before removing duplicates:", df.shape)

df = df.drop_duplicates(subset="track_id").reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
```

## Choosing Audio Features

The 10 audio features below are used to measure similarity between songs. They are converted to `float64`.

Text columns:
- `track_name`, `track_artist`, `track_album_name`: only extra spaces are stripped, the original case is kept because these are shown to the user.
- `playlist_genre`, `playlist_subgenre`: stripped, lowercased and converted to `category` (few unique values).

```python
# 1. Audio features as float
audio_features = ['danceability', 'energy', 'loudness', 'speechiness',
                  'acousticness', 'instrumentalness', 'liveness',
                  'valence', 'tempo', 'duration_ms']

df[audio_features] = df[audio_features].astype('float64')

# 2. Clean text columns
# Names shown to the user: strip spaces only (keep original case)
name_cols = ['track_name', 'track_artist', 'track_album_name']
for col in name_cols:
    df[col] = df[col].str.strip()

# Genre columns: strip + lowercase + category
genre_cols = ['playlist_genre', 'playlist_subgenre']
for col in genre_cols:
    df[col] = df[col].str.strip().str.lower().astype('category')
```

## Visualization

### Summary Statistics

The features are on very different scales:
- `duration_ms` has a mean of about 226,575 and a maximum of 517,810.
- `tempo` has a mean of about 121 and a maximum of about 239.
- `loudness` ranges from about -46 to +1.3.
- The remaining features lie between 0 and 1.

KNN is distance based, so `duration_ms` would dominate the distance. Feature scaling is required (Phase 2).

```python
df[audio_features].describe().T
```

### Histograms

Distribution of each audio feature.

```python
fig, axes = plt.subplots(2, 5, figsize=(20, 7))
fig.suptitle("Audio Feature Distributions", fontsize=16)

for ax, col in zip(axes.ravel(), audio_features):
    ax.hist(df[col], bins=40, color="#1DB954", edgecolor="white")
    ax.set_title(col)
    ax.set_ylabel("Count")

plt.tight_layout()
plt.show()
```

**Observations**
- Right-skewed (most values near 0, long tail to the right): `speechiness`, `acousticness`, `instrumentalness` (the most extreme), `liveness`.
- Left-skewed (long tail to the left): `loudness`, and slightly `danceability` and `energy`.
- `duration_ms` is slightly right-skewed.
- `valence` and `tempo` are roughly balanced.

Scaling fixes the range but not the skew. Whether a transformation is needed will be decided in Phase 2.

### Suspicious Values

Checking for values that are not realistic: a tempo of 0 BPM and songs shorter than 30 seconds.

```python
print("tempo == 0     :", (df["tempo"] == 0).sum())
print("duration < 30s :", (df["duration_ms"] < 30000).sum())
```

### Findings

- Only 5 rows had missing values and were dropped.
- Duplicate `track_id` values were removed: 32,828 rows -> 28,352 unique songs.
- Features have very different scales (`duration_ms`, `tempo`, `loudness`), so scaling is required before KNN.
- Several features are skewed (`speechiness`, `acousticness`, `instrumentalness`, `liveness`, `loudness`).
- Only 1 row has `tempo == 0` and only 2 rows are shorter than 30 seconds. They will be removed in Phase 2.
